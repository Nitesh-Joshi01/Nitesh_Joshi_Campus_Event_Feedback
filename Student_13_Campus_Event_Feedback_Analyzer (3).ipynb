## 1. Import libraries

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_13_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Dataset shape (rows, columns):",df.shape)
print("Column names:",df.columns)
df.info()
print("Missing values in each column:",df.isnull().sum())
print("Number of duplicate rows:",df.duplicated().sum())



```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
Content_array = df['Content_Rating'].to_numpy()

print("overall ratings of content:", round(np.mean(Content_array), 2))
print("Maximum ratings of content::", np.max(Content_array))
print("Minimum ratings of content::", np.min(Content_array))

Organization_array = df['Organization_Rating'].to_numpy()

print("overall ratings of organization:", round(np.mean(Organization_array), 2))
print("Maximum ratings of orgazisation:", np.max(Organization_array))
print("Minimum ratings of orgazisation:", np.min(Organization_array))

```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
df['Total_Rating'] = df['Content_Rating'] + df['Venue_Rating'] + df['Organization_Rating']
df['Average_Rating'] = (df['Total_Rating'] / 3).round(2)


high_ratings = df[df['Average_Rating'] >= 4]

sorted_df = high_ratings.sort_values(by='Average_Rating', ascending=False)
print("Top 5 Feedback Records:")
print(sorted_df.head(5))
```
